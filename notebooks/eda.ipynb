```python
credit = pd.read_csv("../data/processed/credit_clean.csv")

# find the actual extreme rows post-cleaning
print("Most extreme Monthly_Balance values:")
print(credit["Monthly_Balance"].sort_values().head(5))
print(credit["Monthly_Balance"].sort_values().tail(5))

print("\nMost extreme Annual_Income values:")
print(credit["Annual_Income"].sort_values().tail(10))

# now trace one extreme Monthly_Balance row back to its RAW string
worst_balance_idx = credit["Monthly_Balance"].idxmin()
raw = pd.read_csv("../data/raw/train.csv", low_memory=False)
print("\nRaw string behind the worst Monthly_Balance row:")
print(raw.loc[worst_balance_idx, "Monthly_Balance"])

worst_income_idx = credit["Annual_Income"].idxmax()
print("\nRaw string behind the worst Annual_Income row:")
print(raw.loc[worst_income_idx, "Annual_Income"])
```

```python
raw = pd.read_csv("../data/raw/train.csv", low_memory=False)

def clean_numeric_column(series):
    cleaned = series.astype(str).str.replace(r"[^0-9.\-]", "", regex=True)
    return pd.to_numeric(cleaned, errors="coerce")

raw["Annual_Income_clean"] = clean_numeric_column(raw["Annual_Income"])
raw["Monthly_Balance_clean"] = clean_numeric_column(raw["Monthly_Balance"])

# now trace within THIS SAME dataframe, so indices stay aligned
worst_balance_idx = raw["Monthly_Balance_clean"].idxmin()
print("Raw Monthly_Balance string causing the extreme value:")
print(repr(raw.loc[worst_balance_idx, "Monthly_Balance"]))

worst_income_idx = raw["Annual_Income_clean"].idxmax()
print("\nRaw Annual_Income string causing the extreme value:")
print(repr(raw.loc[worst_income_idx, "Annual_Income"]))
```

```python
budget = pd.read_csv("../data/processed/budget_clean.csv")
budget["date"] = pd.to_datetime(budget["date"])

print(f"Date range: {budget['date'].min()} to {budget['date'].max()}")
print(f"Total days spanned: {(budget['date'].max() - budget['date'].min()).days}")
print(f"\nTransactions per category:")
print(budget["category"].value_counts())
print(f"\nTransactions per day (sample):")
print(budget.groupby(budget["date"].dt.date).size().describe())
```

```python
customers = pd.read_csv("../data/processed/customer_clean.csv")

print(f"Total rows: {len(customers)}")
print(f"Unique customers: {customers['Customer ID'].nunique()}")
print(f"Avg transactions per customer: {len(customers) / customers['Customer ID'].nunique():.2f}")

print(f"\nMissing values:")
print(customers.isnull().sum())

print(f"\nCategory distribution:")
print(customers["Category"].value_counts())

print(f"\nTransaction Amount stats:")
print(customers["Transaction Amount"].describe())
```

```python
customers = pd.read_csv("../data/processed/customer_clean.csv")
customers["Birthdate"] = pd.to_datetime(customers["Birthdate"])
from datetime import datetime
customers["Age"] = (datetime.now() - customers["Birthdate"]).dt.days // 365
print(customers["Age"].describe())
```

```python
customers = pd.read_csv("../data/processed/customer_clustered.csv")
print(f"Correlation between Age and Transaction Amount: {customers['Age'].corr(customers['Transaction Amount']):.4f}")

print("\nMean Transaction Amount by Category:")
print(customers.groupby("Category")["Transaction Amount"].mean().sort_values(ascending=False))
```
